## Dataset

[Steel Plate Dataset](http://www.archive.ics.uci.edu/dataset/198/steel+plates+faults)

This dataset contains a table of 20 input attributes that describe the geometry and properties of steels plates, with 7 output attributes, each representing a different type of fault using one-hot encoding.

The goal of this project is to predict the correct type of fault given the input attributes. There is no category for "no fault".

## Implementation Details

There are no missing values in the dataset, although we will have to check for outliers or invalid values. We will also need to convert the one-hot encoded classification labels to a single categorical attribute.

The class distribution of the dataset is not even, about half the labels are "other", while dirtiness and stains both have less than 100 instances. We will need to be careful for false classification of these, and can use stratification to prevent one class from dominating training data.

Since our dataset is mostly continuous, we will have to use binning for some classifiers (e.g. decision trees). We may manually set cut points based on our EDA, or we may do it evenly.

We do not have too many attributes, but if we encounter problems exploring all of them, we may use Principal Component Analysis to reduce the dimensionality of our dataset to something easier to work with.

## Results
Being able to classify the type of fault in a metal plate based on, for instance, computer vision results is very useful as it can save time in industrial settings and provide another figurative "set of eyes" in quality control processes. The primary stakeholder for classifying the faults on metal plates is industry, specifically fields like aerospace or heavy industry where there is a low tolerance for error in regards to the state of raw materials being used to machine parts. Furthermore, the manufacturers of raw steel plates can also use such classification tools to prevent defective parts from being sold to customers.
