## CPSC 322 Final Project Proposal: Bankruptcy Prediction
Spencer Mundel and Damon Lewis

### Dataset
[Bankruptcy Dataset](https://archive.ics.uci.edu/dataset/572/taiwanese+bankruptcy+prediction)

This dataset contains a list of 96 attributes for thousands of different companies based in Taiwan and whether they went legally bankrupt. We will perform principal component analysis at a later time to determine the attributes used.

### Implementation Details
The dataset seems to not have any missing values, which simplifies the cleaning process. However, most of the data is numerical and not categorical, which will require additional steps to use classifiers that need categorical separation (ie. naive bayes or decision trees). Additionally, there are a lot of different attributes which will require principal component analysis to filter down into a more usable dataset.
