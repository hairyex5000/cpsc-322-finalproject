# CPSC 322 Final Project Proposal: Steel Plate Fault Prediction

Spencer Mundel and Damon Lewis

## Dataset

[Steel Plate Dataset](http://www.archive.ics.uci.edu/dataset/198/steel+plates+faults)

This dataset contains a list of 20 input attributes that describe the geometry and properties of steels plates, with 7 output attributes, each representing a different type of fault using one-hot encoding.

The goal of this project is to predict the correct type of fault given the input attributes.
